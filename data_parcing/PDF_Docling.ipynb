```python
from pathlib import Path
from pymupdf import PdfReader, PdfWriter

# Define the source and output file paths
SOURCE_PDF = SOURCE
OUTPUT = Path("./data/ABB_60870.pdf")

# Read the PDF file
reader = PdfReader(SOURCE_PDF)
writer = PdfWriter()

# Ensure the requested page exists in the document
if len(reader.pages) >= 25:
    # Add page 25 (index 24 since it's 0-based)
    writer.add_page(reader.pages[24])
else:
    print(f"The PDF does not contain 25 pages. Total pages: {len(reader.pages)}")

# Save the extracted page to a new PDF file
with OUTPUT.open('wb') as output_pdf:
    writer.write(output_pdf)

print(f"Page 25 has been saved to {OUTPUT}")
```
